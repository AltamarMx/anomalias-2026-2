## El flujo en una línea

```
Mirar  →  ¿Estacionaria?  →  Identificar (p,q)  →  Estimar  →  ¿Residuales blancos?
                                                                 ├── Sí → Pronosticar
                                                                 └── No → Subir orden, repetir
```

Las **dos pruebas** que sostienen todo el método:

| Prueba | ¿Sobre qué? | ¿Qué decide? |
|---|---|---|
| **ADF** | la serie | si puedes empezar |
| **Ljung-Box** | los residuales | si puedes parar |

## Etapa 1 · ¿Tu serie es candidata para ARMA?

**Checklist antes de tocar `ARIMA(...)`:**

- [ ] ¿Tendencia visible? → diferenciar (te vas a **ARIMA**) o desestacionalizar.
- [ ] ¿Estacionalidad determinista (ciclo diario, anual)? → restar climatología, o ir a **SARIMA**.
- [ ] ¿ADF rechaza ($p < 0.05$)? → estacionaria en covarianza ✓
- [ ] ¿Suficientes obs? → mínimo ~50 para AR(1), ~100 para ARMA(2,1), ~200+ con estacionalidad.

> **Recordatorio crítico.** ADF detecta **raíz unitaria**, no estacionalidad. Una serie
> con ciclo diario perfecto puede pasar el ADF y aún así romper ARMA. La verificación
> final es Ljung-Box sobre los residuales, no ADF sobre los datos.

## Etapa 2 · Identificación — leer ACF y PACF

La **tabla espejo** que tienes que memorizar:

| Patrón observado | Modelo a probar |
|---|---|
| ACF decae geométricamente, PACF **corta** en lag $p$ | **AR(p)** |
| ACF **corta** en lag $q$, PACF decae geométricamente | **MA(q)** |
| ACF decae, PACF decae | **ARMA(p,q)** — empezar humilde con (1,1) |
| ACF con picos en lags $k, 2k, 3k, \dots$ | Estacionalidad de período $k$ → SARIMA |
| ACF que decae lentísimo, casi lineal | Posible raíz unitaria → diferenciar primero |

**Regla práctica.** Cuando dudes entre AR y ARMA, empieza por el AR puro y deja
que Ljung-Box decida si necesitas el término MA.

## Etapa 3 · Estimar

```python
from statsmodels.tsa.arima.model import ARIMA

mod = ARIMA(serie, order=(p, 0, q)).fit()
print(mod.summary().tables[1])   # coeficientes y p-values
print(f"AIC = {mod.aic:.1f}   BIC = {mod.bic:.1f}")
```

**Qué mirar en el output:**

- Cada coeficiente debe tener $|z| > 2$ (equivalente a $p < 0.05$).
- Si el último $\phi$ o $\theta$ no es significativo → sobreajustaste, baja orden.
- Coeficientes AR cercanos a 1 → posible raíz unitaria, considera diferenciar.

## Etapa 4 · Diagnosticar

```python
from statsmodels.stats.diagnostic import acorr_ljungbox
from statsmodels.graphics.tsaplots import plot_acf

p, _, q = mod.model.order
resid = mod.resid.iloc[max(p, q):]            # descartar burn-in

plot_acf(resid, lags=20)                      # ¿picos fuera de banda?
lb = acorr_ljungbox(resid, lags=[10, 20], return_df=True)
print(lb)                                     # ¿lb_pvalue > 0.05?
```

**Lectura:**

| Observación | Veredicto |
|---|---|
| ACF residual dentro de bandas + LB $p > 0.05$ | Modelo **adecuado** |
| Picos fuera de banda en ACF residual | Estructura no capturada |
| LB $p < 0.05$ en lag 10 o 20 | Residuales no son blancos |
| Picos en lags 24, 48, 72... (datos horarios) | Estacionalidad diaria ignorada |

## Etapa 5 · Decidir — ¿iterar o parar?

```
                      ┌─────────────────────────┐
                      │   Ljung-Box p > 0.05?   │
                      └────────────┬────────────┘
                                   │
                      ┌────────────┴────────────┐
                      No                        Sí
                      │                         │
              ┌───────┴────────┐       ┌────────┴────────┐
              │ Subir orden:    │       │ ¿Último coef.   │
              │   p+1 o q+1     │       │ significativo?  │
              │ (volver a 3)    │       └────────┬────────┘
              └─────────────────┘                │
                                      ┌──────────┴──────────┐
                                      No                    Sí
                                      │                     │
                             ┌────────┴────────┐    ┌───────┴───────┐
                             │ Sobreajustaste  │    │   ADECUADO    │
                             │   bajar orden   │    │     PARAR     │
                             └─────────────────┘    └───────────────┘
```

**Regla de oro.** Subir orden cuando LB rechaza; parar cuando LB **no** rechaza
Y los nuevos coeficientes son significativos. Si subir orden no apaga la
autocorrelación, el problema **no es de orden** — es de **familia**.

## Selección entre modelos válidos

Si dos o más modelos pasan Ljung-Box, ¿cuál elegir?

| Criterio | Cuándo aplicarlo |
|---|---|
| **Menor BIC** | Producción / parsimonia (BIC penaliza más fuerte los parámetros) |
| **Menor AIC** | Mejor predicción de corto plazo |
| **Menos parámetros** | Empate técnico → siempre el más simple |
| **Coeficientes con $|z| > 2$** | Descartar modelos con parámetros no significativos |

**AIC vs BIC en una frase.** AIC busca el mejor modelo *predictivo*; BIC busca el
modelo *verdadero* (asintóticamente consistente). Para Box-Jenkins en operación,
tiende a usarse BIC por parsimonia.

## Errores comunes — síntoma → diagnóstico → solución

| Síntoma | Diagnóstico | Solución |
|---|---|---|
| AIC bajo **pero** LB rechaza | Modelo mal especificado | Subir orden o cambiar familia |
| LB pasa **pero** último coef. no significativo | Sobreajuste | Bajar orden |
| ACF residual con picos en 24, 48, 72... | Estacionalidad ignorada | SARIMA, o restar climatología |
| $\phi \to 1$ | Cuasi raíz unitaria | Diferenciar → modelar como ARIMA |
| Pronóstico colapsa instantáneamente a la media | Modelo casi sin memoria | Revisar orden, longitud de serie |
| IC del pronóstico no se ensancha | Bug — verificar que `get_forecast` no `predict` | Usar `get_forecast` |
| Convergence warnings al ajustar | Mal condicionado / modelo demasiado complejo | Reducir orden, verificar estacionariedad |
| Coeficientes que cambian de signo entre iteraciones | Multicolinealidad / inestabilidad | Bajar orden |

## Snippet end-to-end

Pipeline completo, listo para adaptar:

```python
import pandas as pd
from statsmodels.tsa.arima.model import ARIMA
from statsmodels.tsa.stattools import adfuller
from statsmodels.graphics.tsaplots import plot_acf, plot_pacf
from statsmodels.stats.diagnostic import acorr_ljungbox

# 1. Cargar y desestacionalizar si hace falta
y = pd.read_csv("...", parse_dates=["time"], index_col="time")["valor"]
clima_h = y.groupby(y.index.hour).mean()
anom = y - y.index.hour.map(clima_h)

# 2. ¿Estacionaria?
stat, p, *_ = adfuller(anom, autolag="AIC")
assert p < 0.05, "No estacionaria — diferenciar o desestacionalizar más"

# 3. Identificar
plot_acf(anom, lags=48)
plot_pacf(anom, lags=48, method="ywm")

# 4-5. Iterar Box-Jenkins
candidatos = [(1,0,0), (2,0,0), (2,0,1), (3,0,1)]
resultados = []
for order in candidatos:
    m = ARIMA(anom, order=order).fit()
    p_, _, q_ = order
    resid = m.resid.iloc[max(p_, q_):]
    lb_p = acorr_ljungbox(resid, lags=[20], return_df=True)["lb_pvalue"].iloc[0]
    resultados.append({"order": order, "AIC": m.aic, "BIC": m.bic, "LB_p": lb_p})
print(pd.DataFrame(resultados))

# 6. Pronóstico (con reincorporación de climatología)
fc = m.get_forecast(steps=24)
pred = fc.predicted_mean + fc.predicted_mean.index.hour.map(clima_h)
ic   = fc.conf_int().add(fc.predicted_mean.index.hour.map(clima_h).values, axis=0)
```

## Frontera de ARMA — cuándo NO usar

ARMA es un martillo elegante pero finito. Estas son las señales de que necesitas
otra herramienta:

| Síntoma | Modelo recomendado |
|---|---|
| ACF residual con picos periódicos en $s, 2s, 3s, \dots$ | **SARIMA** $(p,d,q)(P,D,Q)_s$ |
| Tendencia que no se quita con diferenciación simple | Modelos **estructurales** / espacio-estados |
| Heterocedasticidad evidente (varianza cambia en el tiempo) | **GARCH** sobre residuales |
| Relaciones claramente no lineales | **redes neuronales** / espacio-estados no lineal |
| Variables exógenas relevantes (radiación, presión...) | **SARIMAX** o **VAR** multivariado |
| Cambios de régimen / quiebres estructurales | **Markov-switching** / detección de cambio |
| Necesitas explicar *por qué*, no solo predecir | Modelos **estructurales** o causales |

## Convenciones statsmodels — para no morir en la sintaxis

**`ARIMA(y, order=(p, d, q))`** — el orden son tres números:

- `p` — orden autorregresivo
- `d` — número de diferenciaciones (0 para ARMA puro)
- `q` — orden de media móvil

Equivalencias frecuentes:

| `order=` | Modelo | Cuándo |
|---|---|---|
| `(1, 0, 0)` | AR(1) | inercia simple |
| `(0, 0, 1)` | MA(1) | un solo eco |
| `(2, 0, 1)` | ARMA(2,1) | inercia + cierre con MA |
| `(0, 1, 0)` | I(1) | caminata aleatoria pura |
| `(1, 1, 0)` | ARI(1,1) | tendencia + inercia |
| `(0, 1, 1)` | IMA(1,1) | equivalente a *exponential smoothing* |
| `(1, 1, 1)` | ARIMA(1,1,1) | el caballo de batalla genérico |

**Para simulación con `ArmaProcess`** (notación con polinomio del operador de rezago):

```python
from statsmodels.tsa.arima_process import ArmaProcess

# AR(1):  (1 - phi·L) y_t = eps_t
y = ArmaProcess(ar=[1, -phi], ma=[1]).generate_sample(N)

# MA(1):  y_t = (1 + theta·L) eps_t
y = ArmaProcess(ar=[1], ma=[1, theta]).generate_sample(N)
```

> **Cuidado con el signo.** `ArmaProcess` recibe los **coeficientes del polinomio**
> $1 - \phi L$, no del $\phi$ directo. Por eso es `ar=[1, -phi]` y no `ar=[1, phi]`.
> Es la pregunta-trampa típica del primer encuentro.

## Resumen en 5 líneas

1. **Estaciona** la serie (ADF, climatología, diferenciar).
2. **Identifica** orden tentativo con ACF/PACF (la tabla espejo).
3. **Estima** con `ARIMA(...)` y mira coeficientes + AIC/BIC.
4. **Diagnostica** con Ljung-Box sobre residuales — el único juez.
5. **Itera** subiendo orden hasta blancos; **para** y pronostica.

Si el ciclo no converge, no es problema de orden — es de **familia**. Cambia de modelo.
